## Faktafrågor

### 1. Vad kännetecknar regressionsproblem?
#### Att datan kan predikteras en kunteruerlig beroende värde (y) baserat på en oberoende värde (x) som att prediktera nogåns lön (y) beroende på hens ålder (x).
---

### 2. RMSE, MSE, MAE
- RMSE: En utvärederingsmått som beräkner det minsta felet genom kvadratroten ur medelvärdet av de kvadratiska avvikelserna mellan prediktioner och faktiska värden. Ger ett felmått i samma enhet som y‑variabeln.
- MSE: En utvärederingsmått som beräkna det minsta felet genom kvadraten ur medelvärdet av de kvadratiska avvikelserna mellan prediktioner och faktiska värde. Straffar stora fel extra hårt.
- MAE: En utvärederingsmått som tar medelvärdet av absoluta fel. Mer robust mot outliers.

---
### 3. RMSE vs MSE

#### Nej, rangordningen blir identisk eftersom RMSE är bara kvadratroten av MSE. Om modell A har lägre MSE än modell B, kommer den också ha lägre RMSE.  Skillnaden är endast i skala, inte i jämförelsen.

---
### 4. Gradient descent
#### Det är en optemeringsalgoritm som användes för att hitta en funktions minimum eller maximum. Det görs genom att modellen justerar sina parametrar steg för steg för att minimera en kostnadsfunktion (MSE). Man rör sig “nedför backen” i riktning av den brantaste minskningen tills man når en minimum‑punkt när lutningen blir 0.
---

### 5. Bias–variance trade-off förklarer "Varför vi reguleserar en model"
- Bias: handlar om hur mycket modellen förenklar problemet.
- Varience: handlar om hur känslig modellen är för variationer i datan
* En för komplex modell får låg bias men hög variance blir överanpassning.  
* En för enkel modell får hög bias men låg variance blir underanpassning.  
* Vi reguleserar en model för att hantera och/ eller palansera de två varienter (varians - bias).

### 6. Modellöversikter
- a) Linjär regression: att modellen hittar en rak linje/plan(om den är multilinjär) som bäst passar datan genom att minimera MSE.

- b) Ridge regression: samma som linjär regression men med L2‑regularisering som minskar koefficienternas storlek för att undvika överfitting vilket betyder att det kan finnas oviktiga variabler med.

- c) Lasso regression (L1‑regularisering): den här modellen tenderar till att sätta oviktiga parametrar till 0 vilket gör den populärt då man automatiskt får fram vilka variabler som ska användas i en regressions modell.

- d) Elastic Net: kombinerar L1 och L2 så att den gör både shrinkage och variabelselektion. När mix ratio(gama) "r" är 0 tar vi Ridge regression medan när "r" är 1 tar vi lassoregression.

- e) Support Vector Regression (SVR): den försöker hitta en funktion som ligger inom ett “toleransband” (epsilon) alltså att skapa en väg med en given bredd som innehåller så många observationer som möjligt. Robust mot outliers och kan använda kernels. Den kan användas för både regressionsproblem eller/och klassificeringsproblem.

- f) Beslutsträd (DecisionTreeRegressor): den delar upp datan i regioner genom if‑satser. Enkel att tolka men kan överanpassa därför behöver man regularisera den. Den kan användas för både regressionsproblem eller/och klassificeringsproblem.

- g) Ensemble learning (Voting regression): som kombinerar flera modeller för att få bättre stabilitet och lägre variance bättre än vad enskild modell kan göra.

- h) Random Forest: är ett ensemble (en samling) av många beslutsträd som tränas på olika delar av datan. Den skapas genom bagging även om pasting också förekommer. Ger robusta prediktioner och minskar överfitting därför är det samma som i beslutsträd behövs att regularisera den.
---

### 7. White box vs black box

* White box: modeller som är lätta att tolka (t.ex. linjär regression, beslutsträd).  
* Black box: modeller som är svåra att tolka (t.ex. Random Forest, SVM, neurala nätverk).
---

### 8. Bagging vs pasting

* Bagging: tränar flera modeller på bootstrap samples (sampling med återläggning). Bagging ger mer variation. 
* Pasting: tränar flera modeller på sampling utan återläggning. pasting är mer effektivt vid stora dataset.

### 11. R2-score

#### 
R²-score mäter hur stor del av variationen i den beroende variabeln $y$ som modellen kan förklara.

$$
R^2 = 1 - \frac{SS_{\mathrm{res}}}{SS_{\mathrm{tot}}}
$$

Där:
- $SS_{\mathrm{res}}$ är summan av de kvadrerade residualerna, alltså modellens totala prediktionsfel.
- $SS_{\mathrm{tot}}$ är den totala variationen i de observerade $y$-värdena.

Tolkning:
- **$R^2 = 1$**: modellen förklarar all variation perfekt.
- **$R^2 = 0$**: modellen är ungefär lika bra som att alltid gissa medelvärdet.
- **$R^2 < 0$**: modellen är sämre än att använda medelvärdet som förutsägelse.
