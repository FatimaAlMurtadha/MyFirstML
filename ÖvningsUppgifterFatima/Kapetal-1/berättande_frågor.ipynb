### 1. Hur hänger AI, ML och DL ihop?

1. machine learning is the field of study that gives computers the ability to learn without being explicitly programmed.
2. the size and art of programming computers so they can learn from data.

### 3a. Syftet med tränings-, validerings- och testdata
### 
#### Syftet med tränings validations och testdata är 
- training the data for att träna olika modeller på att se mönster 
- valideringsdata for att utvärdera vilken modell som presenterar bäst genom RMSE.
- test data eftersom dvc en uppskattning på modellens generaliseringsförmåga på ny osedd data.
---
### 3b. Vad är k-delad korsvalidering?
### 
#### en utvärderingsteknik some can vara it's alternative till updelning i träningsdata, valideringsdata och testdata. 
#### att dela upp träningsdata i k lika stora delar (folds) där modellen tränas k gånger. I varje gång används k-1 delar för träning och den återstående delen för validering. Sedan beräknas medelvärde av prestandan av alla samtliga iterationarna.
---
### 3c. Vad är RMSE?
### 
#### är ett standardmått på felet i regressionsmodeller.  det beräknas som kvadratroten ur medelvärdet av de kvadratiska avvikelserna mellan prediktioner och faktiska värden. 
#### Den straffar stora fel eftersom felet dras ur kvadratroten.
$$
\mathrm{RMSE} = \sqrt{\frac{1}{n}\sum_{i=1}^{n}\left(y_i-\hat{y}_i\right)^2}
$$
---
### 3d. Hyperparameter vs parameter
### 
#### Parameter: Värden som modellen lär sig själv från träningsdatan (t.ex. koefficienterna/vikterna w och interceptet b i linjär regression).

#### Hyperparameter: det konst och kunskap eller Inställningar som bestäms manuellt av utvecklaren innan träningen påbörjas för att styra hur inlärningen sker (t.ex. antalet träd i Random Forest, k i k-NN, eller inlärningshastigheten).
---
### 3e. Grid search och refit=True
### 
-  ​Grid search: En systematisk metod (GridSearchCV) som utvärderar alla angivna kombinationer av hyperparametrar med korsvalidering för att hitta den bästa uppsättningen. 
- ​refit=True: En inställning som gör att när den bästa hyperparameterkombinationen hittats, tränar Scikit-learn automatiskt om modellen en sista gång på hela den kombinerade tränings- och valideringsdatamängden.
---
### 3f. Kategorisk data (nominal, ordinal, encoding)
### 
#### 
- ​Nominal: Kategorier utan någon inbördes ordning (t.ex. bilmärke, färg). Omvandlas bäst med One-Hot Encoding. 
-  ​Ordinal: Kategorier som har en naturlig rangordning (t.ex. utbildningsnivå, betyg). Omvandlas bäst med Ordinal Encoding (där ordningen bevaras som tal: 0, 1, 2). 
- ​Encoding: Processen att konvertera icke-numeriska textvärden till numeriska matriser så att algoritmerna kan utföra matematiska beräkningar.
---
### 3g. Vad är feature engineering?
### 
#### Processen att använda domänkunskap för att skapa, transformera, kombinera eller välja ut de mest relevanta variablerna (features) ur rådata för att förenkla för modellen och öka dess prediktionsförmåga.
---
### 3h. Principle of parsimony
### 
#### Även känd som Ockhams rakkniv. Principen innebär att om två modeller uppnår likvärdig prestanda bör man alltid välja den enklare modellen. En enklare modell är lättare att tolka, snabbare att exekvera och har lägre risk för överanpassning.

---

### 4. En modell är en förenkling av verkligheten

#### Modeller filtrerar bort obetydligt brus för att isolera de underliggande mönstren. Som statistikern George Box uttryckte det: "All models are wrong, but some are useful". En modell är en abstraktion av verkligheten designad för ett specifikt syfte, inte en exakt kopia.
---


### 5. Overfitting

#### Överanpassning uppstår när en modell blir för komplex och memorerar träningsdatan, inklusive dess brus och avvikelser. Det resulterar i extremt hög prestanda på träningsdata men dålig prestanda på ny, osett data (dålig generalisering).
---

### 6. Högre är bättre i scikit-learn scoring

#### Scikit-learn följer konventionen att alla score-funktioner fungerar som utvärderingsmått där högre värden innebär bättre resultat (maximering). För förlustfunktioner som MSE (där lägre är bättre) multiplicerar Scikit-learn därför värdet med -1 (t.ex. neg_mean_squared_error).

---

### 7. Tvärsnittsdata, tidsseriedata, paneldata

* ​Tvärsnittsdata (Cross-sectional): Observationer insamlade från flera enheter (t.ex. personer, hus) vid en enkel tidpunkt.
* ​Tidsseriedata (Time-series): Observationer insamlade från en enhet över flera sekventiella tidspunkter (t.ex. aktiekurs minut för minut).
* ​Paneldata (Longitudinal): Kombination av båda – observationer insamlade från flera enheter över flera tidspunkter.

## 9. Varför negative MSE?
#### Optimeringsalgoritmer i Scikit-learn (som GridSearchCV) är byggda för att maximera poängvärden. Eftersom minimering av MSE är det egentliga målet, görs värdet negativt. Då blir att maximera -MSE matematiskt identiskt med att minimera MSE.

### 11. Varför 60–20–20 proportioner?
#### Proportionen 60% träning, 20% validering och 20% test är en klassisk tumregel som säkerställer att:
1. ​Modellen får tillräckligt med data (60%) för att lära sig grundläggande mönster.
2. ​Det finns tillräckligt med data (20%) för obunden jämförelse av hyperparametrar under validering.
3. ​Ett statistiskt tillförlitligt testset (20%) finns kvar som aldrig påverkats av träning eller hyperparameterjustering.
