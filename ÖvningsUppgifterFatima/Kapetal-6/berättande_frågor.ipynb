### 1. Vad är klustring för något? Ge några exempel på tillämpningsområden.

Klustring är en **osupervised learning**‑metod där målet är att gruppera datapunkter baserat på likheter, utan att det finns någon fördefinierad y‑variabel. Modellen försöker hitta naturliga mönster eller grupper i datan.
Exempel på tillämpningar:  
- Kundsegmentering inom marknadsföring. 
- Identifiering av användarbeteenden.
- Gruppering av dokument eller nyhetsartiklar.  
- Bildsegmentering.
- Upptäckt av anomalier.
- Musikrekommendationer baserat på lyssningsmönster.

### 2. Förklara översiktligt hur K-means fungerar. Använd figur 6.3 (sidan 238) och figur 6.4 (sidan 239) i din förklaring. **K‑means** är en algoritm som delar upp datan i *k* kluster genom följande steg:

1. **Sätta igång:**  
   Välj *k* slumpmässiga punkter som startcentroider. Klustringen kan bli dålig om vi väljer fel antal klusterer (figur6.4).

2. **Tilldelning:**  
   Varje datapunkt tilldelas det kluster vars centroid den är närmast.

3. **Uppdatering:**  
   Nya centroids beräknas genom att ta medelvärdet av alla punkter i varje kluster.

4. **Repetition:**  
   Steg 2 och 3 upprepas tills centroids inte längre flyttar sig nämnvärt.

#### Man behöver kolla: 

- **Hur separerade klustren är**  
   tydliga gränser betyder bra klustring.

- **Hur täta klustren är**  
   täta grupper betyder att punkterna liknar varandra.

- **Om klustren överlappar**  
   överlappning kan betyda att *k* är för stort eller att datan inte har tydliga grupper.

- **Formen på klustren**  
  K‑means fungerar bäst när kluster är ungefär runda och lika stora.  
  Om figuren visar avlånga eller oregelbundna kluster kan andra metoder vara bättre.

#### Resultatet blir *k* kluster där varje punkt är nära sin centroid och långt från andra kluster.

#### en K-mean modell kan var bra när formen på klustererna blir sfärsisk, klustererna har en liknande spridning, klustererna är nästan lika store, och när vi specifeara bra antal kluster i samband med modellträningen. På samma sätt har vi en dålig klusermodell i tvärtom fallen.

### 3. Hur kan man välja vilket antal kluster som ska användas för en K‑means modell?  
Använd inertia, silhouette score och silhouette diagram i ditt svar.
 
Val av antal kluster (*k*) är en viktig del av K‑means. Man kan använda:

#### **1. Inertia (Elbow Method)**: Inertia mäter hur nära datapunkterna ligger sina centroids. Man testar flera värden på *k* och ritar en kurva. Den punkt där kurvan “böjer sig” (the elbow) är ofta ett bra val.

#### **2. Silhouette Score**: Mäter hur väl varje punkt passar i sitt kluster jämfört med andra kluster. Score ligger mellan −1 och +1. Högre värde betyder bättre klustring.

#### **3. Silhouette Diagram**: Visar hur varje kluster är strukturerat.  
Bra kluster har:  
- breda och tydliga silhuetter.
- få negativa värden.
- liten överlappning mellan kluster.

Genom att kombinera dessa tre metoder får man en stabil bedömning av det bästa *k*‑värdet.

### 4. Om du kollar på figur 6.10 på sidan 247, hur många kluster hade du valt och varför?  
Är det en “exakt vetenskap” att välja antalet kluster?
 
Figur 6.10 visar en elbow‑kurva där inertia minskar snabbt fram till fyra kluster.  
Det bästa valet av *k* är ofta vid “böjen” — där ytterligare kluster inte ger mycket förbättring.

Men:  
Val av antal kluster är **inte en exakt vetenskap**.  
Det beror på:  
- datans natur.
- syftet med analysen.
- hur klustren ska användas i praktiken.
- om klustren är meningsfulla ur ett affärsperspektiv.

Man kombinerar därför flera metoder och även domänkunskap.

### 5. Hur tolkar man figur 6.13 på sidan 251?
 
Figur 6.13 visar silhouette diagram för olika antal kluster (k=3, 4, 5, 6). Tolkningen handlar om att analysera varje "knivform" (som representerar ett kluster) baserat på två huvudkriterier för att bedöma kvaliteten:

**​Bredden på varje "kniv" (kluster):** 
Bredden på y-axeln representerar antal observationer i klustret. Vi vill ha relativt jämna bredder, vilket indikerar att data är jämnt fördelad. I bilden ser klustren för k=4 mer jämna ut i storlek än för till exempel k=3 (där ett kluster är väldigt stort).

**​Längden på varje "kniv" (Silhouette-koefficienten):**
Längden på x-axeln visar Silhouette-koefficienten för observationerna. Vi vill att de flesta observationerna ska ha en koefficient som överstiger det genomsnittliga Silhouette-scoret (den röda streckade linjen) och sträcker sig så långt mot 1.0 som möjligt. Detta indikerar att observationerna är väl samlade i sina egna kluster och långt ifrån andra kluster.

**​Slutsats:**
Genom att titta på k=4 i figuren ser vi att alla kluster har observationer som sträcker sig förbi den röda linjen och klustren är mer enhetliga i storlek jämfört med andra värden på k. Därför verkar k=4 vara ett bra val även baserat på denna metod.
