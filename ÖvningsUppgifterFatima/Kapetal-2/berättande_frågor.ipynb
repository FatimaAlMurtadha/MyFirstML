### 3. Scikit-learn: principer, estimators, predictors, transformers
* Scikit-learn är ett open scource maskininlärningsbibliotek för python. Det är byggt på några centrala designprinciper vilka är:  
    - 1. KonsistensÖ vilket betyder att alla objekt ska ha ett konsistent gränssnitt och ett begränsatantal metoder.
    - 2. Inspektion: vilket betyder akonstruktorparameterar nås genom att använda .git_param() metoden. Medan lärda parameterar är lagrade och nåbara som publik attribut, vilka nås genom att anvanda ett understräck så här: print(lin_model.coef_).
    - 3. Begränsning av klasser vilket betyder att bara modellalgoritmer som linjärgrossion är representerade genom byggda klasser som Numpy arrays eller SciPy sparas matrices. Medan hyperparametrar representeras när det är möjligt som vanliga python-stänger eller siffror.
    - 4. Rimliga standardvärden för hyperparameterna vilket betyder att scikit-learn specificerat rimliga standardvärden så att man inte behöver förstå teoretiska delar som i många fall inte har någon praktiskt betydelse.
    
* ​Estimators: Objekt som lär sig från data genom metoden .fit(X, y). 
*  ​Transformers: Estimators som transformerar datamängder genom metoden .transform(X) eller .fit_transform(X) (t.ex. StandardScaler, OneHotEncoder). 
*  ​Predictors: Estimators som gör förutsägelser på ny data via metoden .predict(X) (t.ex. LinearRegression, RandomForestRegressor).

## Resonemangfrågor

### 5. Kalle vs Stina – testdata
#### Stina säger rätt. Testdata ska hållas helt avskilt i ett "variabel" tills den slutgiltiga modellen är vald. Kalle riskerar dataläckage (data leakage) genom att utvärdera modellaverväganden mot testdata under utvecklingens gång, vilket gör hans slutgiltiga utvärdering opålitlig och överanpassad.

### 6. Varför misslyckas ML-projekt?
1. Brist på behövlig data.
2. Dåliga modeller.
3. Sakna rätt kompetens och resurser.
4. ​Dålig datakvalitet "Shit in - shit out".
5. Data leakage eller överanpassning.
