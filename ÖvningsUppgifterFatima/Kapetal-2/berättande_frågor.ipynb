### 2. Produktionssättning
#### Processen att ta en AI-modell i bruk i en produktionsmiljö, vilket kan bland annat omfattar tre centrala delar:
* Enhetstestning: Skapa tester för att säkerställa modellens funktionalitet före och under drift.
* Övervakning och omträning: Bevaka prestandan kontinuerligt och träna om modellen regelbundet (t.ex. veckovis eller månadsvis) för att motverka prestandaförlust när underliggande data förändras.
* Spara modellen: Lagra den tränade modellen med bibliotek som joblib eller pickle för att kunna återanvända den direkt utan att behöva träna om den vid varje användning.
---

### 3. Scikit-learn: principer, estimators, predictors, transformers
*  ​Estimators: Objekt som lär sig från data via metoden .fit(X, y). 
*  ​Transformers: Estimators som transformerar datamängder via metoden .transform(X) eller .fit_transform(X) (t.ex. StandardScaler, OneHotEncoder). 
*  ​Predictors: Estimators som gör förutsägelser på ny data via metoden .predict(X) (t.ex. LinearRegression, RandomForestRegressor).
