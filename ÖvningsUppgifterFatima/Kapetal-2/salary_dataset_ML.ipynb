#### 10. Datasetet “salary_dataset.csv” innehåller 29 observationer av personer. YearsExperience är hur många år de arbetat, och Salary är deras lön.

```python
# Import bibliotek

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_validate
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import root_mean_squared_error
```

# a) Läs in datasetet “salary_dataset.csv” med pandas read_csv()-funktion och dela upp i X och y. Avgör själv vilken variabel som ska vara den beroende variabeln y.


```python
salary_data = pd.read_csv("salary_dataset.csv")
salary_data.info()
```

```python
salary_data.head()
```

#### x , y

```python
# YearsExperience är den oberoende variabeln (x)
x = salary_data.drop(columns=["Salary"])
# Salary är den beroende variabeln som vi ska prediktera
y = salary_data["Salary"]
```

# b) Dela upp datasetet i träning- och testset. (Inget validerings-set alltså!)

```python
x_train , x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print("Train size:", x_train.shape)
print("Test size:", x_test.shape)
```

# c) Träna två regressionsmodeller med k-delad korsvalidering med cross_validate()-funktionen från scikit-learn. Använd neg_root_mean_squared_error som scoring. Välj själv hur många iterationer den ska göra genom hyperparametern cv.

```python
# Använding av cross_validation() i training setet
# Linjär regression
lin_reg = LinearRegression()
lin_reg_cv = cross_validate(
    lin_reg,
    x_train,
    y_train,
    scoring="neg_root_mean_squared_error",
    cv=5
)
lin_reg_cv_rmse = -lin_reg_cv["test_score"].mean()

# träna modellen
lin_reg.fit(x_train, y_train)

# Decision tree
tree_reg = DecisionTreeRegressor(random_state=42)
tree_reg_cv = cross_validate(
    tree_reg,
    x_train,
    y_train,
    scoring="neg_root_mean_squared_error",
    cv=5
)
tree_reg_cv_rmse = -tree_reg_cv["test_score"].mean()

# träna
tree_reg.fit(x_train, y_train)

cv_rmse = {
    "Linear Regression": lin_reg_cv_rmse,
    "Decision Tree": tree_reg_cv_rmse
}

print(f"Linear Regression mean CV RMSE: {lin_reg_cv_rmse}")
print(f"Decision Tree mean CV RMSE: {tree_reg_cv_rmse}")
```


# d) Utvärdera modellen som presterar bäst på testsetet. 
* I vårt fall är det Linjär reggrassion.

```python
# Retrain the selected model on all available training data before testing it.
lin_reg.fit(x_train, y_train)
print(np.mean(y_test))
print()
test_pred = lin_reg.predict(x_test)
test_rmse = root_mean_squared_error(y_test, test_pred)

print(f"Test RMSE på den valde modellen är: {test_rmse}")
```

#### Det syns så tydligt att den här modellen är dålig för att den är överanpassad. Den kommer att prediktera väldigt dåligt på någon osedd eller ny data.
